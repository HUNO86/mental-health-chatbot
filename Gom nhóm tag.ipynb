# Code 3:

```python
import json

def optimize_intents_dataset(input_file, output_file):
    # Đọc file dataset gốc
    with open(input_file, 'r', encoding='utf-8') as f:
        data = json.load(f)

    # 1. Định nghĩa Mapping: Chỉ gom nhóm Giao tiếp, Hành động và Cảm xúc cốt lõi
    merge_mapping = {
        # KHẨN CẤP & CẦU CỨU
        "crisis_emergence": ["suicide", "faq_1337085", "death"], 
        
        # GIAO TIẾP & ĐIỀU HƯỚNG
        "smalltalk_welcome": ["greeting", "morning", "afternoon", "evening"],
        "smalltalk_goodbye": ["night", "goodbye", "thanks", "pandora-useful", "user-meditation", "done"],
        "bot_info": ["about", "skill", "creation", "name", "location"],
        "user_frustration": ["hate-you", "hate-me", "wrong", "stupid", "understand"],
        "ask_for_help": ["help", "user-advice"],
        "smalltalk_jokes": ["jokes"],
        
        # HÀNH ĐỘNG
        "action_agree": ["user-agree", "learn-more", "meditation"],
        "action_deny": ["no-approach", "not-talking"],
        "control_switch_topic": ["something-else", "ask", "neutral-response"],
        "control_repeat": ["repeat"],
        
        # CẢM XÚC THỂ HIỆN
        "emo_depression": ["sad", "worthless", "depressed"],
        "emo_anxiety": ["anxious", "stressed", "scared", "problem"],
        "emo_positive": ["happy"],
        "physical_symptom": ["sleep"],
        "social_symptom": ["friends"]
    }

    # 2. Đảo ngược dictionary để tra cứu nhanh: old_tag -> new_tag
    reverse_mapping = {}
    for new_tag, old_tags in merge_mapping.items():
        for old_tag in old_tags:
            reverse_mapping[old_tag] = new_tag

    merged_intents_dict = {}
    independent_intents = []

    # 3. Phân loại và Xử lý dữ liệu
    for intent in data['intents']:
        old_tag = intent.get('tag', '')
        
        if old_tag in reverse_mapping:
            # Nếu tag nằm trong danh sách cần gom -> Merge
            new_tag = reverse_mapping[old_tag]
            if new_tag not in merged_intents_dict:
                merged_intents_dict[new_tag] = {"tag": new_tag, "patterns": [], "responses": []}
            
            merged_intents_dict[new_tag]['patterns'].extend(intent.get('patterns', []))
            merged_intents_dict[new_tag]['responses'].extend(intent.get('responses', []))
        else:
            # Nếu tag KHÔNG nằm trong mapping (toàn bộ nhóm FAQ, Fact) -> Giữ nguyên độc lập
            independent_intents.append(intent)

    # 4. Loại bỏ trùng lặp (duplicates) cho các nhóm đã gom
    for tag, content in merged_intents_dict.items():
        # Dùng set() để lọc trùng, sau đó ép kiểu về list
        content['patterns'] = list(set(content['patterns']))
        content['responses'] = list(set(content['responses']))

    # 5. Tổng hợp lại thành danh sách hoàn chỉnh
    final_intents = list(merged_intents_dict.values()) + independent_intents
    final_json = {"intents": final_intents}

    # 6. Xuất ra file JSON mới
    with open(output_file, 'w', encoding='utf-8') as f:
        json.dump(final_json, f, indent=4, ensure_ascii=False)
        
    print(f"Hoàn thành! Đã xử lý {len(data['intents'])} tags cũ thành {len(final_intents)} tags tối ưu.")
    print(f"File mới đã được lưu tại: {output_file}")

# Thực thi hàm
if __name__ == "__main__":
    # Đảm bảo bạn đang để file input cùng thư mục với script này
    input_filename = 'merged_intents.json'
    output_filename = 'optimized_intents.json'
    
    optimize_intents_dataset(input_filename, output_filename)
```

# Code 4:

```python
import json

# 1. Định nghĩa mapping (quy hoạch) cho các tag Giao tiếp & Cảm xúc
# Các tag FAQ/Fact không có mặt trong danh sách này sẽ tự động được giữ nguyên.
merge_mapping = {
    "crisis_emergence": ["suicide"], 
    "smalltalk_welcome": ["greeting", "morning", "afternoon", "evening"],
    "smalltalk_goodbye": ["night", "goodbye", "thanks", "pandora-useful", "user-meditation", "done"],
    "bot_info": ["about", "skill", "creation", "name", "location"],
    "user_frustration": ["hate-you", "hate-me", "wrong", "stupid", "understand"],
    "ask_for_help": ["help", "user-advice"],
    "smalltalk_jokes": ["jokes"],
    "smalltalk_casual": ["casual", "neutral-response", "no-response", "default"],
    "action_agree": ["user-agree", "learn-more", "meditation"],
    "action_deny": ["no-approach", "not-talking"],
    "control_switch_topic": ["something-else", "ask"],
    "control_repeat": ["repeat"],
    "emo_depression": ["sad", "worthless", "depressed"],
    "emo_anxiety": ["anxious", "stressed", "scared", "problem"],
    "emo_positive": ["happy"],
    "emo_grief": ["death"],
    "physical_symptom": ["sleep"],
    "social_symptom": ["friends"]
}

# Đảo ngược dictionary để tra cứu nhanh từ tag cũ -> tag mới (kỹ thuật tối ưu tốc độ)
reverse_mapping = {}
for new_tag, old_tags in merge_mapping.items():
    for old_tag in old_tags:
        reverse_mapping[old_tag] = new_tag

# 2. Hàm xử lý logic gom nhóm
def merge_intents_data(original_data):
    merged_dict = {}
    
    for intent in original_data["intents"]:
        old_tag = intent["tag"]
        
        # Kiểm tra xem tag này có thuộc nhóm cần gom (đã định nghĩa ở trên) hay không
        if old_tag in reverse_mapping:
            # Nếu có, gán cho nó cái tên tag mới (ví dụ: sad -> emo_depression)
            target_tag = reverse_mapping[old_tag]
        else:
            # Nếu không (như fact-1, faq_123...), giữ nguyên tên gốc
            target_tag = old_tag
            
        # Nếu tag này chưa tồn tại trong danh sách mới, tạo khung cho nó
        if target_tag not in merged_dict:
            merged_dict[target_tag] = {
                "tag": target_tag,
                "patterns": [],
                "responses": []
            }
            
        # Đổ dữ liệu patterns và responses vào tag tương ứng
        merged_dict[target_tag]["patterns"].extend(intent["patterns"])
        merged_dict[target_tag]["responses"].extend(intent["responses"])
        
    # Loại bỏ các câu bị trùng lặp (nếu có) bằng hàm set()
    for tag in merged_dict:
        # Ép kiểu set để xóa trùng, sau đó ép ngược lại thành list để lưu JSON
        merged_dict[tag]["patterns"] = list(set(merged_dict[tag]["patterns"]))
        merged_dict[tag]["responses"] = list(set(merged_dict[tag]["responses"]))
        
    # Trả về format chuẩn của file JSON ban đầu
    return {"intents": list(merged_dict.values())}

# 3. Chạy thực thi file
if __name__ == "__main__":
    # Đọc file dataset gốc của bạn
    with open('merged_intents.json', 'r', encoding='utf-8') as f:
        original_dataset = json.load(f)
        
    # Xử lý gom nhóm
    optimized_dataset = merge_intents_data(original_dataset)
    
    # Xuất ra file mới để đưa vào huấn luyện mô hình
    with open('optimized_intents.json', 'w', encoding='utf-8') as f:
        json.dump(optimized_dataset, f, ensure_ascii=False, indent=4)
        
    print(f"Hoàn tất! Số lượng intent gốc: {len(original_dataset['intents'])}")
    print(f"Số lượng intent sau khi tối ưu: {len(optimized_dataset['intents'])}")
```
