```python
import pandas as pd
import numpy as np

url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv"
df = pd.read_csv(url)

print(df.head(10))

print("\nMissing values (%):")
print(df.isnull().sum() / df.count() * 100)

print(df.dtypes)
```

# TASK 1: Calculate the number of launches on each site


```python
launch_site_counts = df['LaunchSite'].value_counts()
print(launch_site_counts)
```

# TASK 2: Calculate the number and occurrence of each orbit:

```python
orbit_counts = df['Orbit'].value_counts()
print(orbit_counts)
```

# TASK 3: Calculate the number and occurrence of mission outcomes per orbit type

```python
landing_outcomes = df['Outcome'].value_counts()

for i, outcome in enumerate(landing_outcomes.keys()):
    print(i, outcome)

bad_outcomes = set(landing_outcomes.keys()[[1, 3, 5, 6, 7]])
print("\nBad_outcomes (According to slide):")
print(bad_outcomes)
```

# TASK 4: Create a landing outcome label from Outcome column

```python
landing_class = [0 if outcome in bad_outcomes else 1 for outcome in df['Outcome']]
```

```python
df['Class'] = landing_class

print("8 dòng đầu của cột Class:")
print(df[['Class']].head(8))

success_rate = df['Class'].mean()
print(f"\nSum of successfully landing rate: {success_rate:.2%}")
```

```python
df.to_csv("dataset_part_2.csv", index=False)
print("Đã xuất file dataset_part_2.csv thành công!")
```
